## Problema - Crecimiento de subscriptores en una plataforma digital.

## 1) Descripción del problema

Una plataforma de streaming esta analizando el crecimiento de nuevos subscriptores.

Se ha observado que el numero de nuevos subscriptores en el mes n sigue el siguiente comportamiento:

 -En el mes 0 se registraron 0 nuevos usuarios.

 -En el mes 1 se registro un nuevo usuario.

A partir del mes 2, el numero de nuevos usuarios es igual a la suma de los nuevos usuarios de los dos meses anteriores, debido al efecto combinado de referidos del mes anterior y recomendaciones acumuladas del mes previo.


Este problema se basa en la famosa secuencia de fibonacci, en la que se toma los elementos anteriores de una lista para obtener el siguiente elemento, para un programa que no tenga memoizacion puede ser sencillo calcular los primeros terminos de la secuencia, sin embargo si nos vamos un poco mas lejos, si un programa intanta calcular por ejemplo el elemento 1000 de la secuencia este podria tardar horas, dias o hasta semanas, consumiria muchos recursos y es realmente ineficiente, usar memoizacion es acertado, ya que nos permite guardar datos nuevos que en caso de que nos los vuelvan a pedir el programa ya no tendra que recalcularlo, y respondera en un tiempo significativamente mas corto.


## 2) Requerimientos

### 5) Análisis de complejidad 


La complejidad del programa cuando no usamos memoizacion es exponencial(O(n^2)), ya que al no tener un diccionario o un cache donde guardar los resultados previos cada nuevo calculo de subs generara ramas secundarias que ya habia calculado, en cambio al usar memoizacion(Top-down) la complejidad del programa sera lineal(O(n)),ya que tendra guardado en un diccionario o cache los calculos previos de subs, y podra revisar si ya los tenia guardados, de esa forma reutilizar ese dato y a medida que el diccionario va guardando nuevos calculos el programa programa podra responder mas rapido.

Es importante aclarar que si usamos meoizacion, es decir, vamos guardando los datos en un diccionario, estaremos sacrificando almacenamiento del equipo a cambio de que el algoritmo responda de forma mas eficiente.
